Tuples:A tuples is an ordered collection of items that is immutable(cannot be changed after creation)


```python
name=("Vaibhav","Banana")
print(name)
Name=("Vaibhav",1,2,3,False,1.3827)
print(Name)
```

1:Once a tuple is created ,you cannot change,add or remove its element.

```python
# fruit=("Apple","Banana")
# fruit[0]="Orange"
# print(fruit)  #This will show Type Error
#............
#Accessing Element (Indexing)
fruit=("Apple","Banana","Mango")
print(fruit[1])
print(fruit[0])
#NEgative Indexing:
print(fruit[-1])
#.......
```

```python
#Important:
#Creating a single element Tuple 
#Important :A comma is required
a=(5,)
print(a)
print(type(a))
b=(5)
print(b)
print(type(b))
```

Tuple Method:
Since tuple is immutable ,they have only 2 built-in method

```python
# count()
number=(1,2,2,2,2,3,4)
print(number.count(2))
print(number.count(3))
#.......
#Index():
fruit=("apple","banana","mango")
print(fruit.index("apple"))
print(fruit.index("mango"))
```
