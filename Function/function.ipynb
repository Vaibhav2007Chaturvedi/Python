Function:A function is a resuable block of code that perform a specific task.
Instead of writing the same code again and again,we create a function once and call whenever needed.

```python
def greet():
    print("Hello World")

greet()
greet()
greet()

#Function Defination:
def avg():
    a=int(input("Enter a number first: "))
    b=int(input("Enter a number second: "))
    c=int(input("Enter a number third: "))
    average=(a+b+c)/3
    print(average)

avg()
print("Thank You!")
```

```python
def goodDay(name):
    print("Hello "+name)
goodDay("Vaibhav")
goodDay("Ram")
#......
def Hn(name,ending):
    print("Hello"+name+" "+ending)
Hn("Vaibhav","Happy")
Hn("Ram","")
#.........
def Hm(name,ending):
    print("Hello"+" "+name)
    print(ending)
Hm("Vaibhav","Thank You")
Hm("Ram","Thankss")
```

Return in a function:return send a value back to the function

```python
def add(a,b):
    return a+b
add(10,20)
```

```python
def goodDay(name,ending):
    print("Hello"+name)
    print(ending)
a=goodDay("Vaibhav","Bye!")
print(a)
#....
def goodDay(name,ending):
    print("Hello"+name)
    print(ending)
    return "ok"
a=goodDay("Vaibhav","Bye!")
print(a)
#......
def greet(name):
    gr="Hello "+name
    return gr
a=greet("Vaibhav")
print(a)
```
