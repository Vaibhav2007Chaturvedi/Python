Dictionary:A dictionary is a collection of key-value pair.
1:mutuable(cannot be changed)
2:ordered
3:Key must be unique
4:Value can be duplicated
5.written using{}

```python
student = {
    "name": "Vaibhav",
    "age": 19,
    "cgpa": 8.81
}
print(student)
#  key:name and value:vaibhav as it store in this form key:value
```

```python
#Accessing Dictionary Values:we can access the value its key
student={
    "name":"Vaibhav",
    "age" :19
}
print(student["name"])
print(student["age"])
```

Dictionaries Method:

```python
#1: get() Method:We can also access values using get()
student={
    "name":"vaibhav",
    "age":19,
    21:12,
    31:"Vaibhav"
}
print(student.get("name"))
print(student.get("age"))
print(student.get(21))
print(student.get(31))
```

```python
#2:Keys():Gives all keys
student={
    "name":"Vaibhav",
    "age":19,
    "Course":"cse"
}
print(student.keys())

#......
#3:values():Gives all values.
print(student.values())
#....

#4.items():Give both value and keys.
print(student.items())
#NOTE: This is most useful in loop
for key ,value in student.items():
    print(key,value)
```

```python
#5.Update():add new data or change existing data.
#Channging existing value:
student={
    "name":"Vaibhav",
    "age":19
}
# student.update("age":20)  #SyntaxError: invalid syntax  as use do not use {} this
student.update({"age":20})

print(student)

#Add a new value.
student.update({"city":"Delhi"})
print(student)

```

```python
#6.pop():Remove a specific key-value  pair
student={
    "name":"vaibhav",
    "age":19
}
student.pop("age")
print(student)
```

```python
#7.popitem():Remove the last inserted key-value pair.
student={
    "name":"Vaibhav",
    "age":19,
    "course":"Cse"
}
student.popitem()
print(student)


a=student.popitem()
print(a)
#.........
#8.clear():Remove element from dictionary
student.clear()
print(student)

stud={
    "name":"vaibhav"
}
a=stud.clear()
print(a)
```

```python
#9.copy():create a copy dictionary
student={
    "name":"Vaibhav",
    "Age":19,
    "Course":"Cse"
}
new_student=student.copy()
print(new_student)

#...........


#10.setdefault():it is a method used to get a value if the key exists,otherwise add 
# the key with a default value.
#if key already there:
student={
    "name":"vaibhav",
    "age":19
}
x=student.setdefault("name","Rahul")
print(x)
print(student)
#if key do not exist:
x=student.setdefault("city","Delhi")
print(x)
print(student)
stud={}
# x=student("name","") #Error :TypeError
#print(x)
#print(student)
```
