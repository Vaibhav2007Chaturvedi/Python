List:
Python list are containers to store a set of value of any data type
1:list are mutuable (can be changed) as you see friend[0]="Ram"

```python
friend=["Apple","rahul",1,2,89.9,False]
print(friend)
print(friend[0])
friend[0]="Ram"  
print(friend)
print(friend[0])
print(friend[0:4])
```

List Method:

```python
#append():
str1=["Ram","Vaibhav",1,2,3.6,False]
str1.append("Hello")
print(str1)
#............
#insert():insert an element at specific index.
str1.insert(1,"Banana")
print(str1)
#............
#extend():
a=[1,2,3,"ram"]
b=[9.3,7,8,"Hello"]
a.extend(b)
print(a)
```

```python
#remove():Removes the first occurrence of the specified value.
number=[10,20,30,40]
number.remove(20)
print(number)
#.........
#pop():Removes and returns an element by index. If no index is given, it removes the last element.
number.pop()
print(number)
number.pop(1)
print(number)
# number.pop(1): this will indexerror
# print(number)
#.............
#clear():Remove all the element of list
s=["Ram",1,2,3,3,4,4,4,4,4,4]
s.clear()
print(s)
```

```python
# index():
fruit=["banana","Apple","Mango"]
a=fruit.index("Apple")
print(a)  #or
print(fruit.index("Apple"))
#...........
#count():count how many times a value appear
number=[1,2,3,3,3,3,3,3,3,3,4,5]
print(number.count(3))
#.......
#sort():Sorts the list in ascending order.
num=[5,4,3,2,1]
num.sort()
print(num)
print(num.sort())  #it will give none
#as above will show ascending order
#if you want descending order
num.sort(reverse=True)
print(num)
```

```python
#reverse():
number=[1,2,3]
number.reverse()
print(number)
#......
#copy():
a=number.copy()
print(a)
```
