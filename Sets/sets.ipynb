Sets:A sets is a collection of unique elements.
Main Properties:
1:Mutable:can add/remove eleement.
2:Unique:duplicated are automaticlly removed.
3:No indexing.
4:Unordered
5:Can store differnt datatype
6:Use {}.

```python
numbers={1,2,2,2,3,4,5,5,5}
print(numbers)
name={1,"Name",False,2.32}
print(name)
x={}
print(type(x))
x=set()
print(type(x))
```

Sets Method:

```python
#1:add():Adds one element.
numbers={1,2,3,4}
numbers.add(5)
print(numbers)
#...
#2:update():Adds Multiple Element.
number={1,2,3}
number.update([4,5,6])
print(number)
#NOTE:
#  add()       → one element
#  update()    → multiple elements
#........
#3:remove():remove specific element
num={1,2,3}
num.remove(3)
print(num)
#num.remove(9)#This will give error as element do not exist
#print(num)
#.........
#discard():Also remove an element
numb={1,2,3,4}
numb.discard(3)
print(numb)
#NOTE:
#remove()   → error if element doesn't exist
#discard()  → no error if element doesn't exist
```

```python
#pop():Remove an arbitrary element from the set.
numbers={10,20,30}
numbers.pop()
print(numbers)
# Don't think of it as "remove the first element." Sets don't have a fixed order.
numbers.pop()
print(numbers)
numbers.pop()
print(numbers)

#.......
#clear():Remove all element.
number={1,2,3}
number.clear()
print(number)
#......
#copy():Create a copy of the set.
A={1,2,3}
B=A.copy()
print(B)

#......
#union():Combines all element from both sets.
#intersection:Return common value
num1={1,2,3}
num2={4,5,6}
num3={3,4,5}
print(num1.union(num2))
print(num1.intersection(num2))
print(num1.intersection(num3))
#.............
#difference():Return element that are in A but not in B.
C={2,3,4,8,9}
D={2,3,4,5,6}
print(D.difference(C))
```
